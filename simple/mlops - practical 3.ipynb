##  MLOPS - PRACTICAL 3

***

## aim:

To understand the role of AWS services in an MLOps workflow.

**Exam note:** AWS implementation on the PC is not required. AWS may be asked as a viva question.

## problem statement:

Explain how AWS cloud services can support the storage, computation, deployment, monitoring and automation stages of an MLOps workflow.

***

## theory:

### - AWS

Amazon Web Services (AWS) is a cloud platform that provides infrastructure and managed services.

### - amazon S3

S3 is an object-storage service used to store datasets, trained models and other ML artifacts.

### - amazon EC2

EC2 provides virtual computing instances that can be used to run applications, training jobs or inference services.

### - IAM

Identity and Access Management controls identities, authentication, authorization and permissions for AWS resources.

### - cloudwatch

CloudWatch provides monitoring, metrics and logs for AWS resources and applications.

### - codebuild

CodeBuild is a managed build service that can install dependencies, run commands and perform build/test stages.

### - codedeploy

CodeDeploy automates application deployment to supported compute environments.

### - codepipeline

CodePipeline connects stages of a CI/CD workflow such as source, build/test and deployment.

***

## mlops workflow:

**Source Code → Build/Test → Model/Application Artifact → Deployment → Monitoring**

Typical AWS roles:

- **S3** → store datasets and model artifacts
- **EC2** → provide compute/hosting
- **CodeBuild** → build and test
- **CodeDeploy** → deploy
- **CodePipeline** → automate the workflow
- **CloudWatch** → monitor
- **IAM** → control access

***

## viva questions:

1. **What is AWS?**  
   A cloud platform providing infrastructure and managed services.

2. **What is S3?**  
   AWS object storage.

3. **What is an S3 bucket?**  
   A container for storing S3 objects.

4. **Is S3 a database?**  
   No. It is an object-storage service.

5. **What is EC2?**  
   A service providing virtual compute instances.

6. **What is IAM?**  
   AWS identity and access-management service.

7. **What is CloudWatch?**  
   A monitoring and logging service.

8. **What is CodeBuild?**  
   A managed service for building and testing applications.

9. **What is CodeDeploy?**  
   A service that automates application deployment.

10. **What is CodePipeline?**  
    A service used to automate stages of a CI/CD pipeline.

11. **Difference between S3 and EC2?**  
    S3 provides storage; EC2 provides compute.

12. **Why is cloud infrastructure useful in MLOps?**  
    It provides scalable compute, centralized storage, automation and deployment infrastructure.

***

## result:

AWS services can support an MLOps lifecycle by providing storage, computing, access control, build automation, deployment and monitoring. No AWS implementation is required for the practical exam according to the announced exam pattern.

## observation:

AWS services can support different stages of an MLOps workflow. S3 provides object storage, EC2 provides compute, IAM manages permissions, CodeBuild supports build and test stages, CodeDeploy supports deployment, CodePipeline connects CI/CD stages, and CloudWatch supports monitoring and logging. AWS implementation is not required on the examination PC.
