## 1. importing libraries

```python
# NLTK is used for tokenization, stopwords, stemming, lemmatization and POS tagging
import nltk
import re

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("averaged_perceptron_tagger")

from nltk.tokenize import sent_tokenize, word_tokenize, RegexpTokenizer
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
```

## 2. data handling

```python
# Sample text; the same flow can be applied to any document or text column
text = 'He was running and eating at the same time. He has eaten his meal. \nThe quick brown fox jumps over the lazy dog.'

print(text)
```

## 3. sentence tokenization

```python
# Split the document into individual sentences
sentences = sent_tokenize(text)

print(sentences)
```

## 4. word tokenization

```python
# Split each sentence/document into individual word and punctuation tokens
words = word_tokenize(text)

print(words)
```

## 5. regex based tokenization

```python
# Regex lets us select a particular pattern from text
sample = "Hello NLP 2026! My age is 25."

print("Words:", re.findall(r"\w+", sample))
print("Numbers:", re.findall(r"\d+", sample))
print("Non-digits:", re.findall(r"[^0-9]", sample))
```

## 6. regex tokenizer

```python
# RegexpTokenizer uses a pattern to decide what becomes a token
tokenizer = RegexpTokenizer(r"\w+")

print(tokenizer.tokenize(sample))
```

## 7. stopword removal

```python
# Stopwords are common words that may carry less useful information
stop_words = set(stopwords.words("english"))

filtered_words = [
    word for word in words
    if word.lower() not in stop_words
]

print(filtered_words)
```

## 8. stemming

```python
# Stemming reduces words using rule-based suffix removal
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in words
]

print(stemmed_words)
```

## 9. lemmatization

```python
# Lemmatization returns a dictionary/base form using linguistic information
lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("better", pos="a"))
print(lemmatizer.lemmatize("running", pos="v"))
print(lemmatizer.lemmatize("went", pos="v"))
print(lemmatizer.lemmatize("studies", pos="v"))
```

## 10. pos tagging

```python
# POS tagging assigns a grammatical tag to each token
tagged_words = pos_tag(words)

print(tagged_words)
```

## 11. noun phrase chunking

```python
# Define a simple grammar for noun phrases: optional determiner + adjectives + noun
grammar = "NP: {<DT>?<JJ>*<NN>}"

chunk_parser = nltk.RegexpParser(grammar)
tree = chunk_parser.parse(tagged_words)

print(tree)
```

```python
#alternate method 1 to see tree
# Display the parsed noun-phrase tree
tree.pretty_print()
```

```python
#alternate method 2to see tree
for subtree in tree.subtrees():
    if subtree.label() == "NP":
        print(" ".join(word for word, tag in subtree.leaves()))
```

## 12. combined preprocessing

```python
# A compact preprocessing flow: lowercase → tokenize → remove stopwords/punctuation
clean_tokens = [
    word.lower()
    for word in word_tokenize(text)
    if word.isalpha() and word.lower() not in stop_words
]

print(clean_tokens)
```

## viva questions:

1. **What is tokenization?**  
   Splitting text into smaller units called tokens.

2. **Difference between sentence and word tokenization?**  
   Sentence tokenization splits text into sentences; word tokenization splits text into words/punctuation tokens.

3. **What is regex?**  
   A pattern-matching technique used to find or manipulate specific text patterns.

4. **What does `\w+` mean?**  
   One or more word characters.

5. **What does `\d+` mean?**  
   One or more consecutive digits.

6. **What does `[^0-9]` mean?**  
   Any character that is not a digit.

7. **What is a stopword?**  
   A common word that is often removed during preprocessing when it provides limited task-specific information.

8. **Stemming vs lemmatization?**  
   Stemming uses simpler suffix-based reduction and may produce non-dictionary words; lemmatization aims to return valid base forms.

9. **Why specify `pos` during lemmatization?**  
   The correct lemma can depend on the word's grammatical role.

10. **What is POS tagging?**  
    Assigning a part-of-speech tag such as noun, verb, adjective, or adverb to each token.

11. **What is chunking?**  
    Grouping tagged words into meaningful phrases such as noun phrases.

12. **What does `isalpha()` do?**  
    It checks whether a string contains only alphabetic characters.

13. **Why convert text to lowercase?**  
    To reduce unnecessary differences such as `Apple` and `apple`.

14. **Can stopword removal ever be harmful?**  
    Yes. Words such as `not` can affect meaning, so removal should depend on the task.
