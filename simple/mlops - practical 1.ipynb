## aim: 

to train a Random Forest classification model for telecom tower hardware-failure prediction, evaluate the model, save the trained model, and use it to predict failure for a new tower observation.

## problem statement:

Given `tower_telemetry.csv`, use the eight telemetry features to predict `Failure_Within_48Hrs`. Split the data into training and testing sets, train a Random Forest classifier, calculate accuracy, save the model as `telecom_tower_model.pkl`, and use the saved model for a new prediction.

***

### 1. training the model

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import joblib
import json

df = pd.read_csv("tower_telemetry.csv")

X = df[
    [
        "Power_Consumption_W",
        "Temperature_C",
        "Signal_Strength_Percent",
        "Fan_Speed_RPM",
        "Battery_Voltage",
        "Tower_Age_Years",
        "Humidity_Percent",
        "Traffic_Load"
    ]
]

y = df["Failure_Within_48Hrs"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

model = RandomForestClassifier(n_estimators=100, random_state=42)

model.fit(X_train, y_train)

accuracy = model.score(X_test, y_test)

print("Accuracy:", accuracy)

joblib.dump(model, "telecom_tower_model.pkl")

metrics = {"accuracy": accuracy}

with open("metrics.json", "w") as f:
    json.dump(metrics, f, indent=4)

print("Training Completed Successfully")
```

### 2. prediction using the saved model

```python
import joblib
import pandas as pd

model = joblib.load("telecom_tower_model.pkl")

new_data = pd.DataFrame({
    "Power_Consumption_W": [3256],
    "Temperature_C": [42.6],
    "Signal_Strength_Percent": [71],
    "Fan_Speed_RPM": [1745],
    "Battery_Voltage": [47.21],
    "Tower_Age_Years": [9],
    "Humidity_Percent": [65],
    "Traffic_Load": [4301]
})

prediction = model.predict(new_data)

if prediction[0] == 1:
    print("Hardware Failure Predicted")
else:
    print("Tower is Healthy")
```
