## Regex

Regex as a filter:

Suppose you have a box of mixed objects (letters, numbers, symbols). Regex lets you say:

"Give me only the words."
"Find all numbers."
"Remove punctuation."
"Find every email address."

```python
print('\n')   # prints an actual newline (blank line)
print(r'\n')  # prints the literal characters \ and n   r --> raw string
```

```python
import re # regular expressions

text = "Hello123!"

print(re.findall(r'\w', text)) # findall function --> 2 inputs a pattern and a text
```

\w matches one letter, one number, or an underscore at a time.

```python
text = "Hello NLP 2025!"

print(re.findall(r'\w+', text))
```

\w+ matches complete words

```python
text = "My age is 25."

print(re.findall(r'\d', text))
```

matches one digit

```python
text = "Room 123 costs 450 rupees."

print(re.findall(r'\d+', text))
```

+ joins consecutive digits together to form complete numbers.

```python
text = "Cat sat on the mat. 123"

print(re.findall(r'.', text))
```

The dot (.) matches any character (except a newline).

```python
text = "cat likes banana"

print(re.findall(r'[abc]', text))
```

[abc] means "match either a, b, or c."

```python
text = "NLP2026 Practical Session 1"

print(re.findall(r'[^0-9]', text))
```

^ inside square brackets means NOT. So [^0-9] means "anything except a digit."

## Spacy Tokenizer

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
t = "Apple is coming up with new manufacturing hub for iphones in India which will be completed by 2025. The total investment is going to be $100 billion"
```

```python
for ent in nlp(t).ents:
    print(ent.text, "->", ent.label_)
    print(ent.label_, str(spacy.explain(ent.label_))) # the entity, it's explaination
```

doc.ents named entities in the whole doc

```python
for token in nlp(t):
    print(f"{token.text:12} {token.pos_:8} {token.tag_:6} {token.lemma_:12} stop={token.is_stop}")
```

token.textraw word

token.pos_coarse POS (NOUN, VERB...)

token.tag_detailed POS (NN, VBD...)

token.lemma_base form

token.is_stopis it a stopword

```python
clean = [token.lemma_ for token in nlp(t) if not token.is_stop and not token.is_punct]
print(clean)
```

# Stemming and Lemmatization

- Stemming can produce non-dictionary forms (e.g., "running" becomes "run" or "runn").
- Lemmatization gives proper dictionary words by considering the context (e.g., "running" becomes "run", but "better" becomes "good").

## Stemming

```python
from nltk.stem import PorterStemmer # importing class from module
from nltk.tokenize import word_tokenize
```

```python
ps = PorterStemmer()
```

```python
ps.stem("studies")
```

```python
text = "He was running and eating at the same time. He has eaten his meal."
```

```python
# Tokenize the text
words = word_tokenize(text)

# Apply stemming
stemmed_words = [ps.stem(word) for word in words]

print(stemmed_words)
```

## Lemmatization

```python
nltk.download('wordnet') # electronic dictionary
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("better", pos="a"))
print(lemmatizer.lemmatize("running", pos="v"))
print(lemmatizer.lemmatize("went", pos="v"))
print(lemmatizer.lemmatize("studies", pos="v"))
```

Here, we're specifying the part of speech of the word and providing it as arguement.

```python
nltk.download('averaged_perceptron_tagger')         # trained machine learning model
```

Perceptron – a simple type of machine learning classifier (predicts a category — here, a POS tag — based on input features)

Averaged – a training trick that makes the perceptron's predictions more stable/accurate (averages the weights learned across many training steps instead of just using the final ones)

Tagger – its job is to assign a tag to each word (NN, VBD, JJ, etc.)

```python
from nltk import pos_tag
from nltk.corpus import wordnet
def get_pos(word):
    """ Map POS tag to the first character accepted by WordNetLemmatizer """
    print(word)

    tag = pos_tag([word])[0][1][0].upper()
    tag_dict = {"J": wordnet.ADJ, "N": wordnet.NOUN, "V": wordnet.VERB, "R": wordnet.ADV}
    return tag_dict.get(tag, wordnet.NOUN)
```

J → adjective tags start with J (JJ, JJR, JJS) → wordnet.ADJ
N → noun tags start with N (NN, NNS, NNP) → wordnet.NOUN
V → verb tags start with V (VB, VBD, VBG...) → wordnet.VERB
R → adverb tags start with R (RB, RBR, RBS) → wordnet.ADV

```python
text = "He was running and eating at the same time. He has eaten his meal."
```

```python
words = word_tokenize(text)
lemmatized_words = [lemmatizer.lemmatize(word, get_pos(word)) for word in words]
print(lemmatized_words)
```

# POS Tagging

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag # pos_tag is a function which performs POS tagging
```

```python
text = "He was running and eating at the same time. He has eaten his meal."
```

```python
words = word_tokenize(text)
```

```python
print(pos_tag(words))
```

```python
grammer = "NP: {<DT>?<JJ>*<NN>}"
# NP refers to noun phrase
# DT is a determiner ('the','a')
# JJ zero or more adjectives
# NN a single noun
chunk_parser = nltk.RegexpParser(grammer)
tree = chunk_parser.parse(pos_tag(words))
print(tree)
```

```python
tree.pretty_print()
```

```python
words = word_tokenize("The quick brown fox jumps over the lazy dog")
tree = chunk_parser.parse(pos_tag(words))
tree.pretty_print()
```
